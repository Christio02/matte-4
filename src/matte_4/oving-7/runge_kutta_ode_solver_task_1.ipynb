```python
import numpy as np

## a)

def ode_solver(ode, iv, h, N):
    iv = np.array(iv)
    sol = np.zeros((iv.size, N + 1))
    sol[:, 0] = iv
    t0 = 0.0
    tt = np.zeros(N + 1)
    for n in range(N):
        tn=t0 + n * h
        yn=sol[:,n]
        k1 = ode(tn, yn)
        k2 = ode(tn +h/3, yn + h/3 * k1)
        k3 = ode(tn + (2 * h/3), yn + (2 * h)/3 * k2)
        sol_n=yn + h/4 * (k1 +3 * k3)
        sol[:, n + 1] = sol_n
        tt[n + 1] = tn +h
    return sol, tt



## b)

def func(t):
    return t - 1 + 2*np.exp(-t)

def ode(t, y):
    return -y + t

def find_convergence_order():
    N_0 = 4
    iv = [1]
    errors = []
    for i in range(1, N_0 + 1): # check 4 halvings
        N = N_0 * 2 ** i
        h = (1 - 0) / N
        # unsure next
        
        sol, tt = ode_solver(ode, iv, h, N)
        final_val = sol[0, -1]
        error = np.abs(func(1.0) - final_val)
        errors.append((h, error))
    
    for j in range(1, len(errors)):
        h_prev, err_prev = errors[j-1]
        h_curr, err_curr = errors[j]
        p = np.log(err_prev / err_curr) / np.log(h_prev / h_curr)
        
        print(f"h: {h_curr:.4f}, Error: {err_curr:.2e}, Estimated order p: {p:.2f}")
find_convergence_order()
        
        
        
        
        

    
```
