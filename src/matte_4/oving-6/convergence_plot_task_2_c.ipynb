## Convergence plot

- Convergence plot of error for numerical differentiation scheme $$u'(x) \approx \frac{-3u(x) + 4u(x + h) - u(x + 2h)}{2h}$$ when $u(x)=e^{x} \cdot sin(x)$

```python
import numpy as np
import matplotlib.pyplot as plt

def num_diff_scheme(func, step_size, x):
    return (-3 * func(x) + 4 * func(x + step_size) - func(x + 2 * step_size)) / (2 * step_size)


def u(x):
    return np.e**x * np.sin(x)

def u_derivative(x):
    return np.e**x * (np.sin(x) + np.cos(x))

# h = 0.1* 2**-i, from 0 to 6

errors = []
step_sizes = []

x = 0.5
for i in range(7):
    step_size = 0.1 * 2 ** -i
    error = abs(u_derivative(x) - num_diff_scheme(func=u, step_size=step_size, x=x))
    step_sizes.append(step_size)
    errors.append(error)
    
    

log_errors = np.log10(errors)
log_step_sizes = np.log10(step_sizes)
plt.figure()
# plot error
plt.plot(log_step_sizes, log_errors, color ="red")


plt.show()

slope = (log_errors[-1] - log_errors[0]) / (log_step_sizes[-1] - log_step_sizes[0])
print("Observed order:", slope)
```
